# Exploratory Data Analysis (EDA) Project

Analyze the Diabetes dataset using statistical summaries, visualizations, and correlation analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')
```

```python
df=pd.read_csv('eda_diabetes_dataset.csv')
print('Shape:',df.shape)
df.head()
```

```python
print(df.dtypes)
print('\nMissing values:\n',df.isnull().sum())
print('\nDuplicate rows:',df.duplicated().sum())
```

```python
df.describe().T
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['disease_progression'],kde=True)
plt.title('Distribution of Disease Progression')
plt.show()
```

```python
df.hist(figsize=(14,10),bins=20)
plt.tight_layout()
plt.show()
```

```python
corr=df.corr(numeric_only=True)
print(corr['disease_progression'].sort_values(ascending=False))
plt.figure(figsize=(11,8))
sns.heatmap(corr,cmap='coolwarm',center=0)
plt.title('Correlation Heatmap')
plt.show()
```

```python
for feature in ['bmi','bp','s5','s4','s3']:
    plt.figure(figsize=(7,5))
    sns.scatterplot(data=df,x=feature,y='disease_progression')
    plt.title(f'{feature} vs Disease Progression')
    plt.show()
```

## Key Findings

- The dataset contains 442 observations and 10 input features.
- The included dataset has no missing values.
- Statistical summaries show the central tendency and spread of each numerical variable.
- The correlation analysis identifies features with stronger linear associations with the target.
- The heatmap provides an overall view of relationships between numerical variables.
- Scatter plots help inspect selected feature-target relationships.
- Correlation shows association, not causation.
